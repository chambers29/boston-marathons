```python
import os
repo = 'boston-marathons'
counter = 0
max_depth = 50
current = os.getcwd()
while True:
    current_tmp = os.path.basename(os.getcwd())
    if current_tmp == repo:
        print(f"Main dir found: {os.getcwd()}")
        break
    else:
        os.chdir("..")
        counter += 1
        if counter >= max_depth:
            print(f"Reached {max_depth} levels up — '{repo}' not found.")
            os.chdir(current)
            print(f"Current path: {current}")
            break
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import math as math
from scripts.mappings import continent_colors
```

```python
df = pd.read_csv('./data/clean/boston_features.csv')
df.head()
```

```python
plt.style.use('seaborn-v0_8-whitegrid')
```

```python
fix, ax = plt.subplots(figsize=(10,5))

# data=df
# create
ax.hist(df['Official Time'], bins='fd', density=True, color='orange', edgecolor='white')

## ax settings
# x
ax.set_xlabel('Official Time [hours]', weight='bold')
ax.set_xlim(1.8,6.5)
ax.set_xticks(np.arange(2,6.1,0.5))
ax.invert_xaxis()
# y
ax.set_ylabel('Density', weight='bold')

## style
ax.grid(True, linestyle='--', alpha=0.5)
ax.set_title('Marathon finish time distribution', weight='bold')
ax.spines[['top','bottom','left','right']].set_visible(False)

# plt.savefig('test/chart_4.png', dpi=300, bbox_inches='tight')
plt.show()
```
