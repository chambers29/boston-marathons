```python
import os
repo = 'boston-marathons'
counter = 0
max_depth = 50
current = os.getcwd()
while True:
    current_tmp = os.path.basename(os.getcwd())
    if current_tmp == repo:
        print(f"Main dir found: {os.getcwd()}")
        break
    else:
        os.chdir("..")
        counter += 1
        if counter >= max_depth:
            print(f"Reached {max_depth} levels up — '{repo}' not found.")
            os.chdir(current)
            print(f"Current path: {current}")
            break
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import math as math
from scripts.mappings import continent_colors
import seaborn as sns
```

```python
df = pd.read_csv('./data/clean/boston_features.csv')
df.head()
```

```python
plt.style.use('seaborn-v0_8-whitegrid')
```

```python
pivot = df.pivot_table(values='Pace',
                       index='Age Group',
                       columns='Continent',
                       aggfunc='mean')

plt.figure(figsize=(10,8))
ax = sns.heatmap(pivot, cmap='YlOrBr', annot=True, fmt=".2f",
            linewidths=0.5, linecolor='white', cbar_kws={'label': 'Pace [min/km]'})

cbar = ax.collections[0].colorbar
cbar.set_label("Pace [min/km]", fontsize=11, fontweight='bold')
cbar.ax.text(1.5, 0.95, "slower", ha='left', va='center', transform=cbar.ax.transAxes)
cbar.ax.text(1.5, 0.1, "faster", ha='left', va='center', transform=cbar.ax.transAxes)

ax.invert_yaxis()
plt.yticks(rotation=0)

ax.set_ylabel('')
ax.set_xlabel('')
plt.setp(ax.get_xticklabels(), rotation=0, ha="center", fontweight='bold', fontsize=10)
plt.setp(ax.get_yticklabels(), rotation=0, ha="right", fontweight='bold')

ax.set_title("Marathon pace by continents and age groups", fontweight='bold', fontsize=15, pad=15)

plt.savefig('charts/heat_continents-pace-age.png', dpi=300, bbox_inches='tight')
plt.show()
```
