#### FEATURES

```python
# PLACES

# Place_Year column
df = df.sort_values(by=["Year", "Official Time"])
df["Place_Year"] = df.groupby("Year").cumcount() + 1

# Place_Global column
df = df.sort_values(by='Official Time').reset_index(drop=True)
df["Place_Global"] = range(1, len(df)+1)
```

```python
# AGE GROUPS
age = df['Age']

bins = list(np.arange(age.min(), 70, 4)) + [age.max() + 1]
labels = [f"{b}–{bins[i+1]-1}" for i, b in enumerate(bins[:-2])] + ["70+"]

df["Age Group"] = pd.cut(age, bins=bins, labels=labels, right=False)
```

```python
# CONTINENTS
continent_map = pd.read_csv("data/mappings/country_to_continent.csv")
df = df.merge(continent_map, on="Country", how="left")
df["Continent"] = df["Continent"].fillna("Unknown")
```
