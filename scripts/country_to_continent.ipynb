```python
import pandas as pd

continent_map = {
    'Europe': ['UKR','NED','ITA','RUS','BEL','BLR','ESP','IRL','LAT','GBR','CRO','GER','SWE',
               'SUI','SVK','DEN','NOR','FIN','POL','FRA','ISL','POR','AUT','LTU','CYP','CZE',
               'SLO','EST','TUR','LUX','ROU','LIE','GRE','SRB','HUN','MLT','AND','SMR','BUL','ALB'],
    
    'North America': ['USA','CAN','MEX','GUA','CRC','DOM','PAN','BER','HON','CAY','ESA','JAM',
                      'VGB','NCA','BAR','GRN','BAH','TCA'],
    
    'South America': ['AHO','BRA','ECU','COL','CHI','PER','ARG','URU','VEN','FLK','TRI','PAR'],
    
    'Asia': ['PAK','QAT','MGL','KUW','JPN','BRN','HKG','MAS','CHN','PHI','SIN','TWN','KOR','TPE',
             'IND','UAE','VIE','ISR','THA','KSA','INA','JOR','OMA'],
    
    'Africa': ['NGR','ETH','KEN','RSA','ZIM','UGA','BDI','EGY','MAR','ALG'],
    
    'Oceania': ['NZL','AUS']
}

rows = []
for continent, countries in continent_map.items():
    for c in countries:
        rows.append((c, continent))

df = pd.DataFrame(rows, columns=["Country", "Continent"])

df.to_csv("data/mappings/country_to_continent.csv", index=False)
```
